```python
# Plot 3: Distribution of 'age'
plt.figure(figsize=(10, 6))
sns.histplot(df['age'], bins=20, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot 4: Relationship between 'age' and 'stroke'
plt.figure(figsize=(8, 6))
sns.boxplot(x='stroke', y='age', data=df)
plt.title('Age Distribution by Stroke Status')
plt.xlabel('Stroke (0=No, 1=Yes)')
plt.ylabel('Age')
plt.show()
```

```python
# Plot 5: Relationship between 'smoking_status' and 'stroke'
plt.figure(figsize=(10, 6))
sns.countplot(x='smoking_status', hue='stroke', data=df, palette='viridis')
plt.title('Stroke Cases by Smoking Status')
plt.xlabel('Smoking Status')
plt.ylabel('Count')
plt.legend(title='Stroke (0=No, 1=Yes)')
plt.show()
```
